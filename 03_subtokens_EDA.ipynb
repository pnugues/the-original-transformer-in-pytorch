# Exploratory Data Analysis of Subtokens

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
with open('wmt14/train.spm.en', 'r') as f_en:
    en = f_en.read().splitlines()
with open('wmt14/train.spm.de', 'r') as f_de:
    de = f_de.read().splitlines()
```

```python
df = pd.DataFrame({'en': en, 'de': de})
```

```python
df
```

```python
df['en'] = df['en'].str.strip().str.split()
```

```python
df['de'] = df['de'].str.strip().str.split()
```

```python
df
```

```python
df['len_de'] = df['de'].str.len()
```

```python
df['len_en'] = df['en'].str.len()
```

```python
df
```

```python
stats = pd.DataFrame({
    'Colonne': ['len_en', 'len_de'],
    'Moyenne': [df['len_en'].mean(), df['len_de'].mean()],
    'Minimum': [df['len_en'].min(), df['len_de'].min()],
    'Maximum': [df['len_en'].max(), df['len_de'].max()],
    'Médiane': [df['len_en'].median(), df['len_de'].median()],
    'Écart-type': [df['len_en'].std(), df['len_de'].std()],
    'Total mots': [df['en'].apply(len).sum(), df['de'].apply(len).sum()]
})

print(stats.round(2))
```

```python
df['len_en'].hist(bins=50, alpha=0.7, label='Anglais')
df['len_de'].hist(bins=50, alpha=0.7, label='Allemand')
plt.xlabel('Nombre de mots par phrase')
plt.ylabel('Nombre de phrases')
plt.legend()
plt.title('Répartition de la longueur des phrases')
plt.show()
```

```python
# Calcul des longueurs
df['len_de'] = df['de'].apply(len)
df['len_en'] = df['en'].apply(len)

# Filtrage les valeurs extrêmes (ex: garder seulement les phrases de moins de 100 mots)
threshold = 200   # on peut ajuster cette valeur (80, 100, 150...)

mask1 = df['len_de'] <= threshold
mask2 = df['len_en'] <= threshold

plt.figure(figsize=(12, 6))

sns.histplot(df['len_de'][mask1], bins=50, alpha=0.7,
             label='De (allemand)', color='blue')
sns.histplot(df['len_en'][mask2], bins=50, alpha=0.7,
             label='En (anglais)', color='orange')

plt.xlabel('Nombre de mots par phrase')
plt.ylabel('Nombre de phrases')
plt.title('Répartition de la longueur des phrases (sans les atypiques)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Statistiques des atypiques
print("Atypiques pour l'allemand :",
      df[df['len_de'] > 200]['len_de'].count(), "phrases")
print("Atypiques pour l'anglais :",
      df[df['len_en'] > 200]['len_en'].count(), "phrases")

# Visualisation avec deux sous-graphiques
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.histplot(df['len_de'][mask1], bins=50, ax=axes[0], color='blue')
axes[0].set_title('Répartition longueur - allemand (sans atypiques)')
axes[0].set_xlabel('Nombre de mots')

sns.histplot(df['len_en'][mask2], bins=50, ax=axes[1], color='orange')
axes[1].set_title('Répartition longueur - anglais (sans atypiques)')
axes[1].set_xlabel('Nombre de mots')

plt.tight_layout()
plt.show()
```

```python
# Statistiques
print("=== Statistiques longueur allemand ===")
print(df['len_de'].describe())

print("\n=== Statistiques longueur anglais ===")
print(df['len_en'].describe())

# Visualisation sans les atypiques extrêmes
plt.figure(figsize=(14, 6))

# On limite à 200 mots pour une bonne lisibilité (on peut ajuster)
max_display = 200

sns.histplot(df['len_de'][df['len_de'] <= max_display],
             bins=50, alpha=0.7, label='allemand', color='blue', kde=True)

sns.histplot(df['len_en'][df['len_en'] <= max_display],
             bins=50, alpha=0.7, label='anglais', color='orange', kde=True)

plt.xlabel('Nombre de mots par phrase')
plt.ylabel('Nombre de phrases')
plt.title('Répartition de la longueur des phrases (limité à 200 mots)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Affichage des atypiques (les phrases très longues)
print(
    f"\nAtypiques pour l'allemand (> 200 mots) : {df[df['len_de'] > 200].shape[0]} phrases")
print(
    f"Atypiques pour l'anglais (> 200 mots) : {df[df['len_en'] > 200].shape[0]} phrases")

# Affichage des phrases les plus longues
print("\nLes trois phrases les plus longues pour l'allemand :")
print(df.nlargest(3, 'len_de')[['len_de', 'de']])
```

```python
# Affichage des phrases les plus longues
print("\nLes trois phrases les plus longues pour l'anglais :")
print(df.nlargest(3, 'len_en')[['len_en', 'en']])
```
